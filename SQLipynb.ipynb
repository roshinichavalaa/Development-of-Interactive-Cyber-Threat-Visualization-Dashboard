### What is SQL?

**SQL (Structured Query Language)** is a standardized programming language used to manage relational databases and perform various operations on the data within them. It's essential for data professionals because it allows you to:

*   **Create and modify database structures** (tables, indexes, views, etc.).
*   **Insert, update, and delete records** in a database.
*   **Retrieve specific data** from a database through queries.
*   **Manage database security** and access permissions.

### Common SQL Commands

Let's go through some of the most fundamental SQL commands with practical examples.

````python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

print("Connected to in-memory SQLite database.")
````

#### 1. `CREATE TABLE`

Used to create a new table in the database. You define the table's name and the columns, along with their data types and any constraints (e.g., `PRIMARY KEY`, `NOT NULL`).

**Syntax:**
```sql
CREATE TABLE table_name (
    column1_name DATATYPE CONSTRAINTS,
    column2_name DATATYPE CONSTRAINTS,
    ...
);
```

````python
# Create a table named 'Students'
create_table_sql = """
CREATE TABLE Students (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    grade REAL
);
"""
cursor.execute(create_table_sql)
conn.commit()

print("Table 'Students' created successfully.")
````

#### 2. `INSERT INTO`

Used to add new rows of data into a table.

**Syntax:**
```sql
INSERT INTO table_name (column1, column2, ...) VALUES (value1, value2, ...);
-- Or, if inserting values for all columns in order:
INSERT INTO table_name VALUES (value1, value2, ...);
```

````python
# Insert data into the 'Students' table
insert_data_sql = """
INSERT INTO Students (id, name, age, grade) VALUES
(1, 'Alice', 10, 5.0),
(2, 'Bob', 11, 4.5),
(3, 'Charlie', 10, 4.8),
(4, 'David', 12, 5.0);
"""
cursor.execute(insert_data_sql)
conn.commit()

print("Data inserted into 'Students' table.")
````

#### 3. `SELECT`

Used to retrieve data from one or more tables. This is arguably the most common and powerful SQL command.

**Syntax:**
```sql
SELECT column1, column2, ...
FROM table_name
WHERE condition
ORDER BY column_name ASC|DESC
LIMIT number;
```

Let's select all data from our `Students` table.

````python
# Select all data from the 'Students' table
select_all_sql = "SELECT * FROM Students;"
cursor.execute(select_all_sql)

# Fetch all results
results = cursor.fetchall()

# Get column names for DataFrame
column_names = [description[0] for description in cursor.description]

df_students = pd.DataFrame(results, columns=column_names)
print("All students:")
display(df_students)


# Select specific columns and filter by condition
select_filtered_sql = "SELECT name, grade FROM Students WHERE age > 10 ORDER BY grade DESC;"
cursor.execute(select_filtered_sql)

results_filtered = cursor.fetchall()
column_names_filtered = [description[0] for description in cursor.description]

df_filtered = pd.DataFrame(results_filtered, columns=column_names_filtered)
print("\nStudents older than 10, ordered by grade:")
display(df_filtered)
````

#### 4. `UPDATE`

Used to modify existing records in a table.

**Syntax:**
```sql
UPDATE table_name
SET column1 = new_value1, column2 = new_value2, ...
WHERE condition;
```

````python
# Update the grade of 'Bob'
update_sql = "UPDATE Students SET grade = 4.7 WHERE name = 'Bob';"
cursor.execute(update_sql)
conn.commit()

print("Grade for Bob updated.\n")

# Verify the update
cursor.execute("SELECT * FROM Students WHERE name = 'Bob';")
updated_bob = cursor.fetchone()

# Displaying the updated data
column_names = [description[0] for description in cursor.description]
df_updated_bob = pd.DataFrame([updated_bob], columns=column_names)
print("Bob's updated record:")
display(df_updated_bob)
````

#### 5. `DELETE FROM`

Used to remove existing records from a table.

**Syntax:**
```sql
DELETE FROM table_name
WHERE condition;
```

**Caution**: If you omit the `WHERE` clause, all records in the table will be deleted!

````python
# Delete a student named 'Charlie'
delete_sql = "DELETE FROM Students WHERE name = 'Charlie';"
cursor.execute(delete_sql)
conn.commit()

print("Charlie's record deleted.\n")

# Verify the deletion by selecting all remaining students
cursor.execute("SELECT * FROM Students;")
remaining_students = cursor.fetchall()

column_names = [description[0] for description in cursor.description]
df_remaining = pd.DataFrame(remaining_students, columns=column_names)
print("Remaining students:")
display(df_remaining)

# Close the connection
conn.close()
print("\nDatabase connection closed.")
````
